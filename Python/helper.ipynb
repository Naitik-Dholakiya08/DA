```python
# Mean
print(df["Sales"].mean())

# Median
print(df["Sales"].median())

# Mode
print(df["Sales"].mode())

# Minimum
print(df["Sales"].min())

# Maximum
print(df["Sales"].max())

# Standard deviation
print(df["Sales"].std())

# Variance
print(df["Sales"].var())

# Count
print(df["Sales"].count())

# Sum
print(df["Sales"].sum())
```

```python
df = df.dropna()
```

```python
df["Sales"] = df["Sales"].fillna(
    df["Sales"].mean()
)
```

```python
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)
```

```python
df["Region"] = df["Region"].fillna(
    df["Region"].mode()[0]
)
```

```python
# Find duplicate records
print(df.duplicated().sum())

# Display duplicate records
print(df[df.duplicated()])

# Remove duplicate records
df = df.drop_duplicates()

# Verify
print(df.duplicated().sum())
```

```python
print(df[df.duplicated(subset=["OrderID"])])
```

```python
# Sales greater than 1000
high_sales = df[df["Sales"] > 1000]

# Sales less than 500
low_sales = df[df["Sales"] < 500]

# Filter a particular region
gujarat_sales = df[df["Region"] == "Gujarat"]

# Multiple conditions
result = df[
    (df["Sales"] > 1000) &
    (df["Region"] == "Gujarat")
]

print(result)
```

```python
# Sort sales ascending
print(df.sort_values("Sales"))

# Sort sales descending
print(df.sort_values("Sales", ascending=False))

# Top 5 sales records
print(df.nlargest(5, "Sales"))

# Lowest 5 sales records
print(df.nsmallest(5, "Sales"))
```

```python
# Unique regions
print(df["Region"].unique())

# Number of unique regions
print(df["Region"].nunique())

# Count each region
print(df["Region"].value_counts())

# Unique records
print(df.drop_duplicates())
```

```python
# Select specific columns
selected = df[
    ["OrderID", "OrderDate", "Region", "Sales"]
]

# Rename columns
df = df.rename(columns={
    "OrderID": "Order_ID",
    "OrderDate": "Order_Date",
    "Sales": "Total_Sales"
})

print(df.head())
```

```python
print(df["Sales"])

print(df[["Sales", "Region"]])
```

```python
df["Region"] = (
    df["Region"]
    .str.strip()
    .str.title()
)
```

```python
# Convert to uppercase
df["Region"] = df["Region"].str.upper()

# Convert to lowercase
df["Region"] = df["Region"].str.lower()

# Remove spaces
df["Region"] = df["Region"].str.strip()

# Replace text
df["Region"] = df["Region"].str.replace(
    "Gujrat", "Gujarat"
)

# Remove special characters from a text column
df["Product"] = df["Product"].str.replace(
    r"[^a-zA-Z0-9 ]", "", regex=True
)
```

```python
print(df[df["Sales"] < 0])

print(df[df["Quantity"] < 0])

print(df[df["Discount"] > 1])
```

```python
df = df[
    (df["Sales"] >= 0) &
    (df["Quantity"] >= 0) &
    (df["Discount"].between(0, 1))
]
```

```python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Sales"] < lower) |
    (df["Sales"] > upper)
]

print(outliers)
```

```python
# Convert date column into datetime
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

# Extract year
df["Year"] = df["OrderDate"].dt.year

# Extract month
df["Month"] = df["OrderDate"].dt.month

# Extract month name
df["Month_Name"] = df["OrderDate"].dt.month_name()

# Extract day
df["Day"] = df["OrderDate"].dt.day

# Extract weekday
df["Day_Name"] = df["OrderDate"].dt.day_name()

print(df.head())
```

```python
# Total price before discount
df["GrossSales"] = (
    df["Quantity"] * df["UnitPrice"]
)

# Discount amount
df["DiscountAmount"] = (
    df["GrossSales"] * df["Discount"]
)

# Final sales
df["Sales"] = (
    df["GrossSales"] - df["DiscountAmount"]
)

print(df.head())
```

```python
# Addition
df["Total"] = df["Sales"] + df["DiscountAmount"]

# Subtraction
df["NetSales"] = df["GrossSales"] - df["DiscountAmount"]

# Multiplication
df["ProductValue"] = df["Quantity"] * df["UnitPrice"]

# Division
df["AveragePrice"] = df["Sales"] / df["Quantity"].replace(0, np.nan)

# Percentage
df["DiscountPercentage"] = df["Discount"] * 100

# Round values
df["Sales"] = df["Sales"].round(2)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

print(region_sales)
```

```python
summary = df.groupby("Region").agg({
    "Sales": ["sum", "mean", "max", "min", "count"],
    "Quantity": ["sum", "mean"]
})

print(summary)
```

```python
result = df.groupby(
    ["Region", "Category"]
)["Sales"].sum()

print(result)
```
